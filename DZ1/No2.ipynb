```python
import random
import math

class Source:
    def __init__(self,
                 Y: float
                ):
        self.Y = Y

    def generate_photon_directions(self,
                                   E: float
                                  ):
        N = int(E * self.Y)
        phi_s = [random.uniform(0.0, 2.0 * math.pi) for _ in range(N)]
        cosTheta_s = [random.uniform(-1.0, 1.0) for _ in range(N)]
        return tuple(zip(phi_s, cosTheta_s))
        # с косинусами потом работать удобнее, нежели переводить их в углы (обычно)

Source1 = Source(1e4)
print(Source1.generate_photon_directions(2/1e4))
```
